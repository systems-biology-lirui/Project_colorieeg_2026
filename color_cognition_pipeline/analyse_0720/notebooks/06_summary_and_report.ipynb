```python
# [Setup]
from pathlib import Path
import sys, json, pandas as pd
ROOT=Path('/home/lirui/liulab_project/ieeg/Project_colorieeg_2026'); PIPE=ROOT/'color_cognition_pipeline'/'analyse_0720'
sys.path.insert(0,str(PIPE)); import config
subject='test001'; result=config.subject_result_dir(subject)
```

```python
# [Collect] One row per subject/target/channel/analysis for later group statistics
tables=[]
for path in result.rglob('*.csv'):
    try:
        frame=pd.read_csv(path); frame.insert(0,'source_table',str(path.relative_to(result))); frame.insert(0,'subject',subject); tables.append(frame)
    except Exception as exc: print('skip',path,exc)
manifest=pd.concat(tables,ignore_index=True,sort=False) if tables else pd.DataFrame()
group_path=config.RESULT_ROOT/'group'/'tables'/'subject_result_manifest.csv'
manifest.to_csv(group_path,index=False); display(manifest.head())
```

```python
# [Report] Generate a conservative result shell; numerical statements come from saved tables
candidate_path=result/'localizer'/'candidate_electrodes_10mm.csv'
n_candidates=len(pd.read_csv(candidate_path)) if candidate_path.exists() else 0
text=f'''# {subject} fMRI Color-Patch—SEEG 分析报告

- 主候选半径：10 mm
- 候选电极记录数：{n_candidates}
- 单被试结论只针对实际覆盖半球，不进行左右差异推断。

## 结论规则

1. Task1 显著：支持目标邻域颜色选择性。
2. Task2 跨水果解码显著：支持颜色知识信息。
3. Task3→Task2 泛化显著：支持感知与知识共享代码。
4. True–False 显著：解释为物体—颜色一致性/冲突，不能单独等同颜色编码。
5. 不显著：写为当前覆盖和功效下未获得证据，不写为表征不存在。
'''
report_path=config.REPORT_ROOT/f'{subject}_color_patch_report.md'; report_path.write_text(text,encoding='utf-8'); print(report_path)
```
