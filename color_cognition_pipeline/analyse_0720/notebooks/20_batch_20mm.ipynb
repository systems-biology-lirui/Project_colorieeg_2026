# 20 · Multi-subject 20-mm batch

All six subjects are inventoried; only subjects with actual 20-mm electrode coverage enter signal analysis. Figures use the shared publication style.

```python
from pathlib import Path
import sys,pandas as pd
ROOT=Path('/home/lirui/liulab_project/ieeg/Project_colorieeg_2026'); PIPE=ROOT/'color_cognition_pipeline'/'analyse_0720'; sys.path.insert(0,str(PIPE)); import config
from utils.batch20 import write_coverage,load_coverage,preprocess_subject,analyse_subject
config.ensure_output_dirs()
coverage=[]
for subject in config.SUBJECTS:
    cand=write_coverage(subject); coverage.append({'subject':subject,'n_candidates_20mm':len(cand),'status':'eligible' if len(cand) else 'not_eligible'})
coverage=pd.DataFrame(coverage); coverage.to_csv(config.BATCH_RESULT_ROOT/'group'/'tables'/'coverage_summary.csv',index=False); display(coverage)
```

```python
eligible=coverage.loc[coverage.n_candidates_20mm>0,'subject'].tolist(); print('eligible subjects:',eligible)
preprocess=[]
for subject in eligible:
    preprocess.append(preprocess_subject(subject,include_hg=True))
pd.DataFrame(preprocess).to_csv(config.BATCH_RESULT_ROOT/'group'/'tables'/'preprocess_status.csv',index=False); display(pd.DataFrame(preprocess))
```

```python
all_status=[]
for subject in eligible:
    result=analyse_subject(subject,modalities=('erp','hg')); all_status.append({'subject':subject,'n_analysis_tables':len(result)})
pd.DataFrame(all_status).to_csv(config.BATCH_RESULT_ROOT/'group'/'tables'/'analysis_status.csv',index=False); display(pd.DataFrame(all_status))
```
